```python
def escreve_pares(qtd, nome_arquivo):
    num = 0
    arquivo = open(nome_arquivo, "w")
    for i in range(qtd):
      arquivo.write(str(num))
      arquivo.write("\n")
      num *= 2
    arquivo.close()

escreve_pares(10, "pares.txt")
```
