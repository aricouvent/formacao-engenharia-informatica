```python
def leitura(vmatr, vmed):
    for i in range(5):
        matr= int(input("Digite a matrícula: "))
        nota1= float(input("Digite a primeira nota: "))
        nota2= float(input("Digite a segunda nota: "))
        media= (nota1+nota2)/2

        vmatr.append(matr)
        vmed.append(media)

def busca(vmatr, elemprox):
    for i in range(5):
        if vmatr[i] == elemprox:
            return i
    return -1

vmatr = []
vmed = []
leitura(vmatr, vmed)
matrpesq = int(input("Digite a matrícula pesquisada ou 0: "))
while matrpesq != 0:
      pos = busca(vmatr, matrpesq)
      if pos ==-1:
         print("Matrícula não encontrada")
      else:
         print("Média: ", vmed[pos])
      matrpesq = int(input("Digite a matrícula pesquisada ou 0: "))

```
